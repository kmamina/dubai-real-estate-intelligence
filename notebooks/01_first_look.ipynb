```python
import pandas as pd

path = "../data/raw/transactions-2026-11-04.csv"
df = pd.read_csv(path, low_memory=False)

print("Rows, columns:", df.shape)
print("\nColumn names and types:")
print(df.dtypes)
```

```python
missing = (df.isna().mean() * 100).round(1).sort_values(ascending=False)
print("Missing values (%):")
print(missing)

print("\nUnique AREA_EN values:", df["AREA_EN"].nunique())
print("\nTransaction groups:")
print(df["GROUP_EN"].value_counts())
```

```python
df["INSTANCE_DATE"] = pd.to_datetime(df["INSTANCE_DATE"], errors="coerce")
print("Date range:", df["INSTANCE_DATE"].min(), "to", df["INSTANCE_DATE"].max())
print("Unparseable dates:", df["INSTANCE_DATE"].isna().sum())
print("Unique TRANSACTION_NUMBER:", df["TRANSACTION_NUMBER"].nunique(), "of", len(df))
```
